# Комп'ютерний практикум №1

**Тема роботи:** Дослідження дискретних марковських процесів прийняття рішень та оцінювання заданих політик.

**Мета роботи:** Навчитися описувати дискретний MDP, будувати матрицю ймовірносетй переходів для заданої політики, досліджувати розподіли станів і обчислювати функцію цінності. Порівняти результати розрахунків із результатами моделювання траєкторій агента.

**Вступ.** У цьому та наступних комп’ютерних практикумах досліджуватимуться два середовища: готове середовище *FrozenLake* бібліотеки *Gymnasium* із власною конфігурацією карти та власне дискретне MDP, створене за допомогою наданих класів. 

У першій роботі основну увагу приділено опису середовищ та оцінюванню *заданих* політик: як змінюється стаціонарний розподіл станів, яким є очікуваний дисконтований дохід і якою є функція цінності. Пошук оптимальної політики розглядатиметься в наступних роботах.


## Завдання для середовища FrozenLake

1. Створіть власну карту FrozenLake розміром $N \times N$,  $5 \le N \le 9$. Розмістіть початкову клітинку, ціль і щонайменше 4 ополонки. Переконайтеся, що від старту до цілі існує маршрут. Створіть середовище з увімкненим ковзанням (`is_slippery=True`) і зобразіть карту з номерами станів.

2. Дослідіть правила переходів. Поясніть, чому вибрана дія не завжди приводить агента до сусідньої клітинки у вибраному напрямку.

3. Задайте дві політики для всіх нетермінальних станів: рівноймовірний вибір дій та власну детерміновану політику, спрямовану на досягнення цілі. Позначте дії детермінованої політики на карті. На цьому етапі не потрібно доводити, що вона оптимальна.

4. Проведіть серію епізодів за кожною політикою. Оцініть частки успішних епізодів, потраплянь в ополонку та епізодів, зупинених через обмеження кількості кроків. Порівняйте результати моделювання з розрахунками та поясніть можливі розбіжності.

5. Для обох політик $\pi$ побудуйте матриці переходів $P^\pi$ та обчисліть розподіли ймовірностей станів через кілька кроків після старту. Визначте ймовірність досягнення цілі та ймовірність потрапляння в ополонку.

6. Обчисліть цінність початкового стану $v_\pi(s_0)$ для обох політик при заданому $0<\gamma<1$. Порівняйте політики за цінністю та за ймовірністю досягнення цілі. Поясніть, чому ці два показники можуть відрізнятися.


## Створення власного марковського процесу прийняття рішень

За допомогою наданих класів створіть власне *Gymnasium*-сумісне дискретне середовище з **2–4 нетермінальними станами** та **2–4 доступними діями**. Середовище може бути епізодичним (із термінальними станами) або неперервним (continuing, функціонує без завершення епізоду). Якщо ви обираєте епізодичний варіант, додайте більше ніж один термінальний стан.

Самостійно визначте зміст станів і дій, початковий стан або розподіл початкових станів, імовірності переходів та винагороди для переходів $(s,a,s')$. Для кожної пари $(s,a)$ «нетермінальний стан — дія» сума ймовірностей усіх можливих результатів має дорівнювати одиниці.

**Передбачте декілька стохастичний переходів:** для деяких пар $(s,a)$ мають бути можливими щонайменше два наступні стани з ненульовими ймовірностями. Також передбачте хоча б один стан, у якому вибір між діями впливає на переходи або винагороди.

Підготуйте графічне представлення станів і переходів та короткий опис правил середовища. Обране середовище використовуватиметься також у наступних комп’ютерних практикумах.



## Завдання для власного середовища

1. Створіть власне MDP відповідно до вимог роботи. Подайте схему станів і переходів, поясніть зміст дій та винагород. Перевірте, що для кожної пари «нетермінальний стан — дія» сума ймовірностей усіх можливих переходів дорівнює одиниці.

2. Задайте **дві політики**: одну детерміновану й одну стохастичну. У стохастичній політиці передбачте випадковий вибір дії принаймні в одному стані. Поясніть, чим відрізняється поведінка агента за цими політиками.

3. Дослідіть довгострокову поведінку середовища за кожної політики:
   - **якщо середовище не має термінальних станів**, знайдіть стаціонарний розподіл, якщо він є єдиним; в іншому разі поясніть, чому результат залежить від початкового стану або політики;
   - **якщо середовище має термінальні стани**, визначте ймовірності потрапляння в кожен із них із заданого початкового стану. Якщо завершення епізоду не гарантоване, зазначте це окремо.

4. Змоделюйте траєкторії агента за обома політиками. Порівняйте результати моделювання з розрахунками: розподіли станів після вибраної кількості кроків, частоти завершення епізоду або відвідування станів, а також оцінку доходу з початкового стану. Для середовища без термінальних станів оцінюйте дохід за достатньо довгою, але скінченною траєкторією.

5. Для кожної політики побудуйте матрицю ймовірностей переходів $P_\pi$. Обчисліть розподіл імовірностей станів після кількох кроків, починаючи з обраного початкового стану або початкового розподілу. Порівняйте отримані розподіли та поясніть відмінності між політиками.

7. Для кожної політики обчисліть функцію цінності $v_\pi(s)$ при заданому коефіцієнті дисконтування $0<\gamma<1$. Порівняйте цінності початкового стану та поясніть їхній зв’язок із переходами й винагородами.

9. Сформулюйте висновок: яка політика має більшу цінність у початковому стані, як відрізняється поведінка агента та наскільки результати моделювання узгоджуються з теоретичними розрахунками.
